```python

import pandas as pd

url = "https://www.football-data.co.uk/mmz4281/2425/E0.csv"

matches = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Number of matches:", len(matches))

matches.head()
```

```python

home_stats = matches.groupby("HomeTeam").agg(
    Home_Goals_Scored=("FTHG", "mean"),
    Home_Goals_Conceded=("FTAG", "mean"),
    Home_Matches=("FTHG", "count")
)

away_stats = matches.groupby("AwayTeam").agg(
    Away_Goals_Scored=("FTAG", "mean"),
    Away_Goals_Conceded=("FTHG", "mean"),
    Away_Matches=("FTAG", "count")
)

print("Home performance:")
display(home_stats.round(2).head())

print("Away performance:")
display(away_stats.round(2).head())
```

```python

team_stats = home_stats.join(away_stats)

team_stats = team_stats.round(2)

display(team_stats.head(10))
```

```python

form_data = matches[
    ["Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR"]
].copy()

form_data["Date"] = pd.to_datetime(
    form_data["Date"], dayfirst=True, errors="coerce"
)

form_data = form_data.dropna(
    subset=["Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR"]
)

form_data = form_data.sort_values("Date").reset_index(drop=True)

display(form_data.head())
```

```python

team_history = {}

for _, match in form_data.iterrows():
    home = match["HomeTeam"]
    away = match["AwayTeam"]

    home_goals = int(match["FTHG"])
    away_goals = int(match["FTAG"])

    if home not in team_history:
        team_history[home] = []

    if away not in team_history:
        team_history[away] = []

    if home_goals > away_goals:
        home_result, away_result = "W", "L"
    elif home_goals < away_goals:
        home_result, away_result = "L", "W"
    else:
        home_result, away_result = "D", "D"

    team_history[home].append({
        "date": match["Date"],
        "opponent": away,
        "goals_scored": home_goals,
        "goals_conceded": away_goals,
        "result": home_result
    })

    team_history[away].append({
        "date": match["Date"],
        "opponent": home,
        "goals_scored": away_goals,
        "goals_conceded": home_goals,
        "result": away_result
    })

print("Teams tracked:", len(team_history))
print("Arsenal matches recorded:", len(team_history.get("Arsenal", [])))
print("Arsenal's last five results:")

for match in team_history["Arsenal"][-5:]:
    print(match["date"].strftime("%d %b %Y"), match["result"],
          f'{match["goals_scored"]}-{match["goals_conceded"]}')
```

```python

def calculate_form(recent_matches):
    points = 0

    for match in recent_matches:
        if match["result"] == "W":
            points += 3
        elif match["result"] == "D":
            points += 1

    return points
```

```python

team_form = {}

for team, history in team_history.items():
    recent_matches = history[-5:]
    points = calculate_form(recent_matches)

    team_form[team] = {
        "Recent_Form": "".join(
            match["result"] for match in recent_matches
        ),
        "Form_Points": points,
        "Goals_Scored_Last_5": sum(
            match["goals_scored"] for match in recent_matches
        ),
        "Goals_Conceded_Last_5": sum(
            match["goals_conceded"] for match in recent_matches
        )
    }

form_table = pd.DataFrame.from_dict(team_form, orient="index")
form_table.index.name = "Team"

display(
    form_table.sort_values("Form_Points", ascending=False)
)
```

```python

full_team_stats["Team_Strength"] = (
    full_team_stats["Overall_Goals_Scored"] * 2
    - full_team_stats["Overall_Goals_Conceded"]
    + full_team_stats["Form_Points"] / 5
)

full_team_stats = full_team_stats.sort_values(
    "Team_Strength", ascending=False
)

display(
    full_team_stats[
        [
            "Overall_Goals_Scored",
            "Overall_Goals_Conceded",
            "Form_Points",
            "Team_Strength"
        ]
    ].round(2)
)
```

```python

def get_pre_match_form(team, past_matches, n=5):
    home_matches = past_matches[
        past_matches["HomeTeam"] == team
    ].copy()

    away_matches = past_matches[
        past_matches["AwayTeam"] == team
    ].copy()

    home_matches["Goals_Scored"] = home_matches["FTHG"]
    home_matches["Goals_Conceded"] = home_matches["FTAG"]

    away_matches["Goals_Scored"] = away_matches["FTAG"]
    away_matches["Goals_Conceded"] = away_matches["FTHG"]

    team_matches = pd.concat(
        [home_matches, away_matches]
    ).sort_values("Date")

    recent = team_matches.tail(n)

    return {
        "Matches": len(recent),
        "Goals_Scored": int(recent["Goals_Scored"].sum()),
        "Goals_Conceded": int(recent["Goals_Conceded"].sum()),
        "Average_Scored": round(recent["Goals_Scored"].mean(), 2),
        "Average_Conceded": round(recent["Goals_Conceded"].mean(), 2)
    }

print("Arsenal:", get_pre_match_form("Arsenal", past_matches))
print("Chelsea:", get_pre_match_form("Chelsea", past_matches))
```

```python

def predict_from_history(home_team, away_team, past_matches):
    home = get_pre_match_form(home_team, past_matches)
    away = get_pre_match_form(away_team, past_matches)

    # Require at least five previous matches per team
    if home["Matches"] < 5 or away["Matches"] < 5:
        return None

    expected_home = (
        home["Average_Scored"] + away["Average_Conceded"]
    ) / 2

    expected_away = (
        away["Average_Scored"] + home["Average_Conceded"]
    ) / 2

    if expected_home > expected_away:
        prediction = "H"
    elif expected_home < expected_away:
        prediction = "A"
    else:
        prediction = "D"

    return {
        "Expected_Home_Goals": expected_home,
        "Expected_Away_Goals": expected_away,
        "Predicted_Result": prediction
    }
```

```python

# Cell 43: Calculate probabilities for possible goal counts

import math

def poisson_probability(expected_goals, actual_goals):
    return (
        math.exp(-expected_goals)
        * expected_goals ** actual_goals
        / math.factorial(actual_goals)
    )

# Example: a team expected to score 1.5 goals
for goals in range(5):
    probability = poisson_probability(1.5, goals)
    print(f"{goals} goals: {probability:.2%}")
```

```python

# Cell 44: Calculate match outcome probabilities

def predict_match_probabilities(home_goals, away_goals, max_goals=10):
    home_win = 0
    draw = 0
    away_win = 0

    for h in range(max_goals + 1):
        home_prob = poisson_probability(home_goals, h)

        for a in range(max_goals + 1):
            away_prob = poisson_probability(away_goals, a)
            score_prob = home_prob * away_prob

            if h > a:
                home_win += score_prob
            elif h == a:
                draw += score_prob
            else:
                away_win += score_prob

    return {
        "Home_Win_%": round(home_win * 100, 2),
        "Draw_%": round(draw * 100, 2),
        "Away_Win_%": round(away_win * 100, 2)
    }


# Test with Arsenal expected to score 1.4
# and Chelsea expected to score 1.1
probabilities = predict_match_probabilities(1.4, 1.1)

print(probabilities)
print("Total probability:", sum(probabilities.values()), "%")
```

```python

# Cell 45: Rank the most likely scorelines

def most_likely_scorelines(home_goals, away_goals, max_goals=5):
    scorelines = []

    for h in range(max_goals + 1):
        for a in range(max_goals + 1):
            probability = (
                poisson_probability(home_goals, h)
                * poisson_probability(away_goals, a)
            )

            scorelines.append({
                "Score": f"{h}-{a}",
                "Probability_%": round(probability * 100, 2)
            })

    scorelines_df = pd.DataFrame(scorelines)
    return scorelines_df.sort_values(
        "Probability_%", ascending=False
    ).head(10).reset_index(drop=True)


display(most_likely_scorelines(1.4, 1.1))
```

```python

# Cell 46: Backtest Poisson predictions

poisson_backtest = []

for i in range(50, len(form_data)):
    match = form_data.iloc[i]
    past_matches = form_data.iloc[:i].copy()

    home = get_pre_match_form(match["HomeTeam"], past_matches)
    away = get_pre_match_form(match["AwayTeam"], past_matches)

    if home["Matches"] < 5 or away["Matches"] < 5:
        continue

    expected_home = (
        home["Average_Scored"] + away["Average_Conceded"]
    ) / 2

    expected_away = (
        away["Average_Scored"] + home["Average_Conceded"]
    ) / 2

    probabilities = predict_match_probabilities(
        expected_home, expected_away
    )

    predicted_result = max(
        probabilities, key=probabilities.get
    )

    # Convert probability dictionary key to H, D or A
    result_code = {
        "Home_Win_%": "H",
        "Draw_%": "D",
        "Away_Win_%": "A"
    }[predicted_result]

    poisson_backtest.append({
        "Date": match["Date"],
        "HomeTeam": match["HomeTeam"],
        "AwayTeam": match["AwayTeam"],
        "Predicted_Result": result_code,
        "Actual_Result": match["FTR"],
        "Correct": result_code == match["FTR"],
        "Home_Win_%": probabilities["Home_Win_%"],
        "Draw_%": probabilities["Draw_%"],
        "Away_Win_%": probabilities["Away_Win_%"]
    })

poisson_backtest_df = pd.DataFrame(poisson_backtest)

display(poisson_backtest_df.head(10))
print("Matches tested:", len(poisson_backtest_df))
```

```python
# Cell 57: Calculate home and away performance before a match

def get_venue_stats(team, past_matches):
    home_matches = past_matches[
        past_matches["HomeTeam"] == team
    ]

    away_matches = past_matches[
        past_matches["AwayTeam"] == team
    ]

    return {
        "Home_Games": len(home_matches),
        "Home_Scored": home_matches["FTHG"].mean(),
        "Home_Conceded": home_matches["FTAG"].mean(),
        "Away_Games": len(away_matches),
        "Away_Scored": away_matches["FTAG"].mean(),
        "Away_Conceded": away_matches["FTHG"].mean()
    }
```

```python
# Cell 58: Venue-aware expected goals

def predict_venue_goals(home_team, away_team, past_matches):
    league_home_goals = past_matches["FTHG"].mean()
    league_away_goals = past_matches["FTAG"].mean()

    home = get_venue_stats(home_team, past_matches)
    away = get_venue_stats(away_team, past_matches)

    if (
        home["Home_Games"] == 0
        or home["Away_Games"] == 0
        or away["Home_Games"] == 0
        or away["Away_Games"] == 0
    ):
        return None

    expected_home = (
        home["Home_Scored"] + away["Away_Conceded"]
    ) / 2

    expected_away = (
        away["Away_Scored"] + home["Home_Conceded"]
    ) / 2

    return {
        "Expected_Home_Goals": expected_home,
        "Expected_Away_Goals": expected_away,
        "League_Home_Avg": league_home_goals,
        "League_Away_Avg": league_away_goals
    }

# Test using matches played before 16 March 2025
test_date = pd.Timestamp("2025-03-16")
past_matches = form_data[form_data["Date"] < test_date].copy()

display(
    predict_venue_goals("Arsenal", "Chelsea", past_matches)
)
```

```python
# Cell 61: Backtest venue-aware Poisson predictions

venue_backtest = []

for i in range(50, len(form_data)):
    match = form_data.iloc[i]
    past = form_data.iloc[:i].copy()

    estimates = predict_venue_goals(
        match["HomeTeam"],
        match["AwayTeam"],
        past
    )

    if estimates is None:
        continue

    probs = predict_match_probabilities(
        estimates["Expected_Home_Goals"],
        estimates["Expected_Away_Goals"]
    )

    predicted = {
        "Home_Win_%": "H",
        "Draw_%": "D",
        "Away_Win_%": "A"
    }[max(probs, key=probs.get)]

    venue_backtest.append({
        "HomeTeam": match["HomeTeam"],
        "AwayTeam": match["AwayTeam"],
        "Actual_Result": match["FTR"],
        "Predicted_Result": predicted,
        "Home_Win_%": probs["Home_Win_%"],
        "Draw_%": probs["Draw_%"],
        "Away_Win_%": probs["Away_Win_%"]
    })

venue_backtest_df = pd.DataFrame(venue_backtest)

print("Matches tested:", len(venue_backtest_df))
display(venue_backtest_df.head())
```

```python
# Cell 62: Evaluate venue-aware probability quality

actual = venue_backtest_df["Actual_Result"]

venue_home = (actual == "H").astype(int)
venue_draw = (actual == "D").astype(int)
venue_away = (actual == "A").astype(int)

venue_brier = (
    (venue_backtest_df["Home_Win_%"] / 100 - venue_home) ** 2
    + (venue_backtest_df["Draw_%"] / 100 - venue_draw) ** 2
    + (venue_backtest_df["Away_Win_%"] / 100 - venue_away) ** 2
).mean()

print(f"Venue-aware Brier score: {venue_brier:.4f}")
print(f"Previous Poisson score: 0.6256")
print(
    "Result:",
    "Venue-aware model improved"
    if venue_brier < 0.6256
    else "No improvement yet"
)
```
